# TEXTVQA

```python
from datasets import load_dataset

ds = load_dataset("lmms-lab/textvqa")
```

```python
ds
```

```python
ds["train"][0]
```

```python
from PIL import Image
from pathlib import Path
import json

path = Path("./textvqa")
path.mkdir(exist_ok=True)
image_path = path / "images"
image_path.mkdir(exist_ok=True)

with open(path / "metadata.jsonl", "w", encoding="utf-8") as f:
    
    for i, ex in enumerate(ds["test"]):
        img = ex["image"]
        img.save(image_path/ f"{i:06d}.jpg", format="JPEG")
        rec = {
            "image_id": f"{i:06d}",
            "question": ex["question"],
            "answers": ex["answers"],
            "ocr_tokens": ex["ocr_tokens"],
        }
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")
```

# DOCVQA eval

```python
from datasets import load_dataset

ds = load_dataset("vikhyatk/docvqa-val")
```

```python
ds['validation'][0]
```

```python
import json
from pathlib import Path


folder = Path("./docvqa_val")
folder.mkdir(exist_ok=True)
out_path = folder / "docvqa_val.jsonl"
image_root = folder / "images"
image_root.mkdir(exist_ok=True)

with open(out_path, "w", encoding="utf-8") as f:
    for idx, sample in enumerate(ds['validation']):
        img = sample["image"]    

        img_filename = f"{idx}.png"
        img_path = image_root / img_filename
        img.save(img_path, format="PNG")

        qa_list = sample["qa"]         # list of {"question","answers"}

        for qid, qa in enumerate(qa_list):
            record = {
                "id": f"{idx}_{qid}",
                "image": img_filename,
                "question": qa["question"],
                "answers": qa["answers"],
            }
            f.write(json.dumps(record, ensure_ascii=False) + "\n")
            
```

# MathVista

```python
from datasets import load_dataset

ds = load_dataset("AI4Math/MathVista")
ds['testmini'][0]
```

```python
path = Path("./mathvista")
path.mkdir(exist_ok=True)

image_path = path / "images"
image_path.mkdir(exist_ok=True)

with open(path / "metadata.jsonl", "w", encoding="utf-8") as f:
    for i, ex in enumerate(ds["testmini"]):
        img = ex["decoded_image"]
        img.save(path / ex["image"], format="PNG")
        rec = {
            "pid": ex["pid"],
            "image": ex["image"],          # 圖片檔名
            "query": ex["query"],          # 給 model 用
            "answer": ex["answer"],        # GT
            "answer_type": ex["answer_type"],
            "precision": ex["precision"],
            "question": ex["question"],      # 給人類看的
        }
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")
```
